### Ensemble Algo

```python
from sklearn.metrics import classification_report
```

```python
def get_prediction(row, threshold=0.5):
    high_prob = row.max()
    if high_prob >= threshold:
        max_class = row.idxmax()
        return max_class
    return "Other"


def classify(threshold,data=meta_data):
    global report_df
    y_true = data['pattern']
    y_pred = data.drop('pattern', axis=1).apply(get_prediction, threshold=threshold, axis=1)
    report = classification_report(y_true, y_pred, output_dict=True, zero_division=0)
    macro = report['macro avg']
    report_df = pd.DataFrame(report).T
    return macro['precision'], macro['recall'], macro['f1-score'],report_df
```

```python
best = None
best_class_wise = pd.DataFrame()
for th in range(0,100,5):
    th = th/100
    precision, recall, f1, report_df = classify(th,meta_data)
    
    best_class_wise[th] = report_df['f1-score']

    if best is None or f1 > best[0]:
        best = (f1, th, precision, recall)
    print(f"Threshold: {th} => Precision: {precision}, Recall: {recall}, F1-Score: {f1}")

print("-"*10)
print(f"Best Threshold: {best[1]} => Precision: {best[2]}, Recall: {best[3]}, F1-Score: {best[0]}")
best_class_wise['best_threshold'] = best_class_wise.idxmax(axis=1)
```
