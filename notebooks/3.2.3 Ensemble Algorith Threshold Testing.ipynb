### Ensemble Algo

```python
from sklearn.metrics import classification_report
```

```python
def get_prediction(row, threshold=0.5):
    high_prob = row.max()
    if high_prob >= threshold:
        max_class = row.idxmax()
        return max_class
    return "None Type"


def classify(threshold):
    y_true = meta_data['pattern']
    y_pred = meta_data.drop('pattern', axis=1).apply(get_prediction, threshold=threshold, axis=1)
    report = classification_report(y_true, y_pred, output_dict=True, zero_division=0)
    macro = report['macro avg']
    return macro['precision'], macro['recall'], macro['f1-score']


classify(0.5)
```

```python
# Debug: inspect true and predicted labels
unique_true = meta_data['pattern'].value_counts()
unique_pred = meta_data.drop('pattern', axis=1).apply(get_prediction, axis=1).value_counts()
print('y_true labels:', unique_true.to_dict())
print('y_pred labels:', unique_pred.to_dict())
```
