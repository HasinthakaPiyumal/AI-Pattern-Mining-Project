```python
import pandas as pd
from sklearn.metrics import classification_report,f1_score
```

```python
old_to_new_mapping = {
    "Advanced LLM Prompting": "Advanced LLM Prompting",
    "Classical Models": "Forecasting with Classical Models",
    "Enhanced User Intent Comprehension with LLMs": "LLM-based User Intent Extraction",
    "LLM Agent Training & Alignment": "LLM Fine-Tuning, Training & Alignment",
    "LLM Context Management": "LLM-based Context Management",
    "LLM Results Evaluation": "Evaluating LLM Results",
    "LLM based Multimodal Generative Prompting": "LLM-based Multimodal Generative Prompting",
    "LLM based Planning, Iterative Optimizations and ReAct, or Reasoning, Think Step by step, XoT":
        "LLM-based Planning, XoT, ReAct, or Reasoning",
    "Model Abstraction Pattern": "Model Abstraction",
    "Modular LLM Agent Architectures": "Agent Architecture",
    "None": "None",
    "Preprocessing Text and Numerical Data": "Preprocessing Text and Numerical Data",
    "Reliable, Transparent, & Augmented LLMs":
        "Enabling Reliability, Explainability, or Robustness",
    "Retrieval Augmented Generation(RAG) Optimization for LLMs":
        "Retrieval Augmented Generation(RAG)",
    "Structured Output & Formatting for LLMs":
        "Performing Structured Output & Formatting for LLMs",
    "Tool Use for LLMs": "Using Tools with LLMs"
}
```

```python
def format_cluster_id(id):
    names = id.split("/")
    return names[-2]+"/"+names[-1]

def format_pattern_name(name):
    return name.strip().split(":")[0]

def map_old_to_new(old_name):
    old_name = old_name.strip()
    return old_to_new_mapping.get(old_name, old_name)
```

```python
labeled_data = pd.read_csv('/home/hasinthaka/Documents/Projects/AI/Pattern Mining/pipeline/data/datasets/labeled_data.csv')
labeled_data_map = [format_cluster_id(i['file']) for i in labeled_data[labeled_data['verified']==True].to_dict(orient='records')]
pattern_name_map = {format_cluster_id(i['file']): i['verified_pattern'] for i in labeled_data.to_dict(orient='records')}
def is_verified(cluster_id):
    formatted_id = format_cluster_id(cluster_id)
    return formatted_id in labeled_data_map
```

```python
def show_report(y_true, y_pred):
    report = classification_report(y_true, y_pred,labels=sorted(set(y_true)), zero_division=0, output_dict=True)
    df = pd.DataFrame(report).transpose()
    return df
```
